```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from prophet import Prophet
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv("./data/cleaned_health_fitness_data.csv")
df["date"] = pd.to_datetime(df["date"], dayfirst=True, errors="coerce")
df = df.dropna(subset=["date"])
```

```python
def prophet_residuals(df, metric):
    p_df = df[["date", metric]].rename(columns={"date": "ds", metric: "y"})
    model = Prophet()
    model.fit(p_df)

    forecast = model.predict(p_df)
    df[f"{metric}_predicted"] = forecast["yhat"]
    df[f"{metric}_residual"] = df[metric] - df[f"{metric}_predicted"]

    return df

df = prophet_residuals(df, "avg_heart_rate")
df = prophet_residuals(df, "hours_sleep")
```

```python
def threshold_anomaly(series, k=3):
    mean = series.mean()
    std = series.std()
    return abs(series - mean) > k * std

df["hr_anomaly_threshold"] = threshold_anomaly(df["avg_heart_rate_residual"])
df["sleep_anomaly_threshold"] = threshold_anomaly(df["hours_sleep_residual"])
```

```python
features = pd.read_csv("./data/tsfresh_features.csv")

scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)

dbscan = DBSCAN(eps=1.2, min_samples=10)
cluster_labels = dbscan.fit_predict(scaled_features)

features["cluster"] = cluster_labels

df["cluster_anomaly"] = df["participant_id"].map(
    lambda x: x in features[features["cluster"] == -1].index
)
```

```python
df["anomaly"] = (
    df["hr_anomaly_threshold"] |
    df["sleep_anomaly_threshold"] |
    df["cluster_anomaly"]
)

df["anomaly_label"] = df["anomaly"].map({True: "Anomalous", False: "Normal"})
```

```python
fig_hr = px.line(
    df, x="date", y="avg_heart_rate",
    title="Heart Rate Anomalies",
    color="anomaly_label"
)

fig_hr.show()

fig_hr.write_image("./visualizations/heart_rate_anomalies.png")
```

```python
fig_sleep = px.line(
    df, x="date", y="hours_sleep",
    title="Sleep Pattern Anomalies",
    color="anomaly_label"
)

fig_sleep.show()

fig_sleep.write_image("./visualizations/sleep_anomalies.png")
```
